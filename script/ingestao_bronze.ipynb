```python
import duckdb
import pandas as pd
import os
from datetime import datetime
```

```python
# criando a conexão com o banco
con = duckdb.connect(database = 'dados_duckdb.db', read_only = False)
```

```python
# carregando os dados
arquivo = 'z0019_2.csv'
data_ingestao = datetime.now()
df = pd.read_csv(f'../landing/{arquivo}', sep = ";")
df['nome_arquivo'] = arquivo
df['data_ingestao'] = data_ingestao
```

```python
df
```

```python
# criando a tabela bronze
con.execute("""
    create table if not exists bronze_z0019(
    NATBR varchar,
    MAKTX varchar,
    WERKS varchar,
    MAINS varchar,
    LABST varchar,
    nome_arquivo varchar,
    data_ingestao timestamp
    )
""")
```

```python
# inserindo os dados do df para a tabela bronze_produtos
con.execute('insert into bronze_z0019 select * from df')
```

```python
# visualizando o resultado da tabela bronze_produtos
resultado = con.execute('select * from bronze_z0019').fetchdf()
resultado
```

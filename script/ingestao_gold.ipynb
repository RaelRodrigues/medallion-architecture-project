```python
import duckdb
```

```python
# criando uma conexão
con = duckdb.connect(database = 'dados_duckdb.db', read_only = False)
```

```python
# carregando os dados da tabela produtos
df = con.execute('select * from produtos').fetchdf()
df
```

```python
# removendo duas colunas
df2 = df.drop(columns=['id_categoria', 'id_fornecedor'])
df2
```

```python
# criando uma nova tabela para gold
con.execute("""
    create table if not exists dim_produtos(
    id_prooduto bigint,
    nm_produto text,
    vl_produto float)
""")
```

```python
# inserindo os dados na tabela
con.execute('insert into dim_produtos select * from df2')
```

```python
# consultando os dados dentro de dim_produtos
df_dim = con.execute('select * from dim_produtos').fetchdf()
df_dim
```

```python
con.close()
```
